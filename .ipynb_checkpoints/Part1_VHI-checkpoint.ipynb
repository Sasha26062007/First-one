### 1: Завантаження даних
Завдання: Для кожної з адміністративних одиниць України завантажити тестові файли з VHI-індексом.Додати дату та час завантаження.Передбачити механізм запобігання повторного завантаження.

```python
import urllib.request
import datetime
import os
import pandas as pd

def download_vhi_data():
    # Створюємо папку для файлів, якщо її немає
    os.makedirs("vhi_data", exist_ok=True)
    
    # Отримуємо поточну дату та час у зручному форматі
    now = datetime.datetime.now().strftime("%Y-%m-%d_%H-%M")
    
    # Область 0 качати не потрібно, тому йдемо від 1 до 27
    for i in range(1, 28):
        # Перевірка на колізії: чи є вже файл для цієї області в папці
        existing_files = [f for f in os.listdir("vhi_data") if f.startswith(f"vhi_id_{i}_")]
        if existing_files:
            print(f"Дані для області {i} вже завантажені ({existing_files[0]}). Пропускаємо.")
            continue
            
        url = f"https://www.star.nesdis.noaa.gov/smcd/emb/vci/VH/get_TS_admin.php?country=UKR&provinceID={i}&year1=1981&year2=2024&type=Mean"
        filename = f"vhi_data/vhi_id_{i}_{now}.csv"
        
        try:
            urllib.request.urlretrieve(url, filename)
            print(f"Успішно завантажено: {filename}")
        except Exception as e:
            print(f"Помилка завантаження області {i}: {e}")

# Викликаємо функцію
download_vhi_data()
```

### 2: Очищення даних та зміна індексів
Завдання: Зчитати файли у pandas dataframe, прибрати зайве, заповнити пропуски.Змінити індекси областей за українською абеткою.

```python
def create_clean_dataframe():
    frames = []
    files = os.listdir("vhi_data")
    
    # Словник: ключ - старий індекс NOAA, значення - новий індекс (за укр. абеткою)
    # Наприклад, 1 (Вінницька) за NOAA була під номером 24.
    index_map = {
        1: 22, 2: 24, 3: 23, 4: 25, 5: 3, 6: 4, 7: 8, 8: 19, 9: 20, 10: 21,
        11: 9, 12: 26, 13: 10, 14: 11, 15: 12, 16: 13, 17: 14, 18: 15, 19: 16,
        20: 27, 21: 17, 22: 18, 23: 6, 24: 1, 25: 2, 26: 7, 27: 5
    }
    
    for file in files:
        if file.endswith(".csv"):
            old_id = int(file.split('_')[2])
            new_id = index_map.get(old_id, old_id)
            
            # Читаємо файл. NOAA дає сміття в 1-му рядку, пропускаємо його (header=1)
            df = pd.read_csv(f"vhi_data/{file}", index_col=False, header=1)
            
            # Очищення даних: видаляємо HTML теги (іноді NOAA їх вставляє в кінці)
            df = df.dropna()
            df = df[~df['year'].astype(str).str.contains('<')]
            
            # Видаляємо пробіли в назвах колонок (data cleaning)
            df.columns = df.columns.str.replace('<br>', '').str.strip()
            df['year'] = df['year'].astype(int)
            
            # Додаємо правильний індекс області
            df['Area_ID'] = new_id
            frames.append(df)
            
    # Збираємо все в одну таблицю
    full_df = pd.concat(frames).drop_duplicates().reset_index(drop=True)
    return full_df

df = create_clean_dataframe()
print("Дані успішно очищено та зведено!")
display(df.head())
```

### 3: Аналіз та вибірки
Завдання: Реалізувати процедуру для формування вибірки: ряд VHI для області за вказаний рік.

```python
def get_vhi_for_year(dataframe, province_id, year):
    # Фільтруємо таблицю за ID та роком
    result = dataframe[(dataframe['Area_ID'] == province_id) & (dataframe['year'] == year)][['week', 'VHI']]
    return result

print("VHI для області 1 (Вінницька) за 2020 рік:")
display(get_vhi_for_year(df, province_id=1, year=2020).head())
```

### 4: Вибірка за діапазон років та кілька областей
Завдання: Ряд VHI за вказаний діапазон років для вказаних областей.

```python
def get_vhi_range(dataframe, province_ids, start_year, end_year):
    # Фільтруємо за списком областей (використовуємо isin) та діапазоном років
    result = dataframe[
        (dataframe['Area_ID'].isin(province_ids)) & 
        (dataframe['year'] >= start_year) & 
        (dataframe['year'] <= end_year)
    ][['year', 'week', 'Area_ID', 'VHI']]
    
    return result

# Передаємо список областей: 1 (Вінницька) та 10 (Київська)
print("VHI для Вінницької (1) та Київської (10) областей за період 2018-2020 років:")
display(get_vhi_range(df, province_ids=[1, 10], start_year=2018, end_year=2020).head(10))
```

### 5: Статистика
Завдання: Пошук екстремумів (min та max) для вказаних областей та років.

```python
def get_extremes(dataframe, province_id, year):
    filtered_df = dataframe[(dataframe['Area_ID'] == province_id) & (dataframe['year'] == year)]
    if filtered_df.empty:
        return "Немає даних"
    
    vhi_min = filtered_df['VHI'].min()
    vhi_max = filtered_df['VHI'].max()

    vhi_mean = filtered_df['VHI'].mean()
    vhi_median = filtered_df['VHI'].median()
    
    return vhi_min, vhi_max, vhi_mean, vhi_median

min_vhi, max_vhi, mean_vhi, median_vhi = get_extremes(df, province_id=1, year=2020)
print(f"Для Вінницької області (ID 1) у 2020 році:\nМінімальний VHI: {min_vhi}\nМаксимальний VHI: {max_vhi}")
print(f"\nСередній VHI: {mean_vhi}\nМедіана: {median_vhi}")
```
